# Dataset 1 — Appliances Energy Prediction (UCI)
## Situação
A empresa de eficiência energética está analisando o comportamento de uma residência de baixo consumo. A equipe deseja identificar períodos de consumo elevado dos eletrodomésticos e observar quais condições de temperatura e umidade estavam presentes nesses momentos.

### 1. Carregue a amostra no Pandas e apresente head(), shape, info() e describe().

```python
df1 = pd.read_csv('/content/drive/My Drive/SERS/EX_01_AMOSTRA.csv')
```

```python
df1.head()
```

```python
print(f"Total de Registros: {df1.shape[0]}")
print(f"Total de Colunas: {df1.shape[1]}")
```

```python
df1.info()
```

```python
df1.describe()
```

```python
df1.columns
```

### 2. Renomeie Appliances para Consumo_Eletrodomesticos e simplifique o nome de pelo menos três atributos ambientais.

```python
df1.rename(columns={
    'Appliances': 'Consumo_Eletrodomesticos',
    'T1': 'Temperatura_1',
    'T2': 'Temperatura_2',
    'T3': 'Temperatura_3'
}, inplace=True)
```

```python
df1.columns
```

### 3. Determine o maior consumo de eletrodomésticos registrado na amostra.

```python
consumo_maximo_d1 = df1['Consumo_Eletrodomesticos'].max()
print(f"Maior consumo de eletrodomésticos: {consumo_maximo_d1:.2f}")
```

### 4. Calcule um limiar correspondente a 70% do valor máximo e crie um DataFrame contendo somente os registros acima desse limite.

```python
limiar_70_d1 = 0.70 * consumo_maximo_d1
df1_consumo_elevado = df1[
    df1['Consumo_Eletrodomesticos'] > limiar_70_d1
].copy()

print(f"Limiar de 70%: {limiar_70_d1:.2f}")
df1_consumo_elevado.head()
```

### 5. Conte os registros selecionados e calcule qual percentual da amostra eles representam.

```python
quantidade_elevada_d1 = len(df1_consumo_elevado)
percentual_elevado_d1 = quantidade_elevada_d1 / len(df1) * 100

print(f"Registros acima do limiar: {quantidade_elevada_d1}")
print(f"Percentual da amostra: {percentual_elevado_d1:.2f}%")
```

### 6. Calcule a temperatura média de T1 e crie outro DataFrame contendo simultaneamente consumo acima de 70% do máximo e temperatura acima da média.

```python
temperatura_media_d1 = df1['Temperatura_1'].mean()
df1_consumo_e_temperatura = df1[
    (df1['Consumo_Eletrodomesticos'] > limiar_70_d1) &
    (df1['Temperatura_1'] > temperatura_media_d1)
].copy()

print(f"Temperatura média de T1: {temperatura_media_d1:.2f}")
print(f"Registros com consumo elevado e temperatura acima da média: {len(df1_consumo_e_temperatura)}")
df1_consumo_e_temperatura.head()
```

### 7. Compare os dois DataFrames e explique o efeito da inclusão da temperatura como segundo critério.

```python
reducao_d1 = quantidade_elevada_d1 - len(df1_consumo_e_temperatura)
percentual_mantido_d1 = (
    len(df1_consumo_e_temperatura) / quantidade_elevada_d1 * 100
    if quantidade_elevada_d1 else 0
)

display(Markdown(f'''
O filtro apenas por consumo selecionou **{quantidade_elevada_d1}** registros. Ao incluir também temperatura de T1 acima da média de **{temperatura_media_d1:.2f}**, restaram **{len(df1_consumo_e_temperatura)}** registros, uma redução de **{reducao_d1}**. Portanto, **{percentual_mantido_d1:.2f}%** dos períodos de consumo elevado também apresentaram temperatura acima da média. A segunda condição torna o recorte mais específico e mostra uma coincidência entre as condições, sem comprovar sozinha uma relação de causa e efeito.
'''))
```

# Dataset 2 — Steel Industry Energy Consumption (UCI)
## Situação
A equipe de gestão de energia de uma indústria siderúrgica deseja localizar situações de consumo elevado e verificar se coincidem com carga elevada ou fator de potência menos favorável.

### 1. Carregue a amostra e renomeie as colunas solicitadas.

```python
ARQUIVO_D2='/content/drive/My Drive/SERS/EX_02_AMOSTRA.csv'
df2=pd.read_csv(ARQUIVO_D2); df2.columns=df2.columns.str.strip()
df2.rename(columns={'Usage_kWh':'Consumo_kWh',
                    'Lagging_Current_Power_Factor':'Fator_Potencia_Atrasado',
                    'Leading_Current_Power_Factor':'Fator_Potencia_Adiantado',
                    'Lagging_Current_Reactive.Power_kVarh':'Potencia_Reativa_Atrasado',
                    'Leading_Current_Reactive_Power_kVarh':'Potencia_Reativa_Adiantada',
                    'WeekStatus':'Status_da_Semana',
                    'Day_of_week':'Dia_da_Semana',
                    'Load_Type':'Tipo_de_Carga'},inplace=True)
df2.columns
```

### 2. Apresente head(), shape, info() e describe().

```python
display(df2.head()); print(f'Registros: {df2.shape[0]} | Colunas: {df2.shape[1]}'); df2.info(); display(df2.describe(include='all'))
```

### 3. Determine o maior consumo e o limiar de 75% do máximo.

```python
max_d2=df2['Consumo_kWh'].max(); limiar_d2=.75*max_d2
print(f'Máximo: {max_d2:.2f} kWh | Limiar de 75%: {limiar_d2:.2f} kWh')
```

### 4. Filtre os registros acima do limiar e calcule quantidade e percentual.

```python
alto_d2=df2[df2['Consumo_kWh']>limiar_d2].copy()
qtd_d2=len(alto_d2); pct_d2=qtd_d2/len(df2)*100
print(f'Quantidade: {qtd_d2} | Percentual: {pct_d2:.2f}%'); alto_d2.head()
```

### 5. Verifique quantos registros do recorte pertencem à categoria Maximum Load.

```python
carga_d2=alto_d2['Tipo_de_Carga'].astype(str).str.strip().str.replace('_',' ',regex=False)
qtd_maxload_d2=carga_d2.str.casefold().eq('maximum load').sum()
print(f'Registros Maximum Load: {qtd_maxload_d2}')
```

### 6. Observe um fator de potência e defina um limite coerente para valores baixos.

```python
COL_FATOR_D2='Fator_Potencia_Atrasado'
display(df2[COL_FATOR_D2].describe())
limite_fator_d2=df2[COL_FATOR_D2].quantile(.25)
print(f'Limite escolhido (1º quartil): {limite_fator_d2:.2f}')
```

**Justificativa:** o primeiro quartil delimita os 25% menores valores observados e fornece um critério relativo e reproduzível. Se houver um limite técnico definido pela equipe, substitua a variável limite_fator_d2.

### 7. Filtre simultaneamente consumo elevado e fator de potência abaixo do limite.

```python
critico_d2=df2[(df2['Consumo_kWh']>limiar_d2)&(df2[COL_FATOR_D2]<limite_fator_d2)].copy()
print(f'Registros no segundo recorte: {len(critico_d2)}'); critico_d2.head()
```

### 8. Explique por que o segundo conjunto merece maior atenção.

```python
display(Markdown(f'''Dos **{qtd_d2}** registros de consumo elevado, **{len(critico_d2)}** também tiveram fator de potência abaixo de **{limite_fator_d2:.2f}**. Eles merecem atenção por combinarem demanda elevada e fator de potência menos favorável, possível sinal de uso menos eficiente e maior potência reativa. O recorte prioriza investigação, mas não comprova a causa.'''))
```

# Dataset 3 — Power Consumption of Tetouan City (UCI)
## Situação
A equipe de distribuição deseja identificar qual das três zonas apresenta o maior pico e observar as condições ambientais nos momentos de maior demanda.

### 1. Carregue a amostra e renomeie as três variáveis de consumo.

```python
ARQUIVO_D3='/content/drive/My Drive/SERS/EX_03_AMOSTRA.csv'
df3=pd.read_csv(ARQUIVO_D3); df3.columns=df3.columns.str.strip()
df3.rename(columns={'Zone 1 Power Consumption':'Consumo_Zona_1',
                    'Zone 2  Power Consumption':'Consumo_Zona_2',
                    'Zone 3  Power Consumption':'Consumo_Zona_3'},inplace=True)
display(df3.head()); df3.columns
```

### 2. Determine o consumo máximo em cada zona.

```python
zonas_d3=['Consumo_Zona_1','Consumo_Zona_2','Consumo_Zona_3']
maximos_d3=df3[zonas_d3].max(); maximos_d3
```

### 3. Identifique a zona com o maior pico.

```python
zona_d3=maximos_d3.idxmax(); pico_d3=maximos_d3.max()
print(f'Zona: {zona_d3} | Pico: {pico_d3:.2f}')
```

### 4. Calcule 70% do máximo da zona identificada e filtre os registros.

```python
limiar_d3=.70*pico_d3; alto_d3=df3[df3[zona_d3]>limiar_d3].copy()
print(f'Limiar: {limiar_d3:.2f}'); alto_d3.head()
```

### 5. Conte os registros e calcule o percentual.

```python
qtd_d3=len(alto_d3); pct_d3=qtd_d3/len(df3)*100
print(f'Quantidade: {qtd_d3} | Percentual: {pct_d3:.2f}%')
```

### 6. Calcule a temperatura média e aplique também essa condição.

```python
temp_media_d3=df3['Temperature'].mean()
alto_temp_d3=df3[(df3[zona_d3]>limiar_d3)&(df3['Temperature']>temp_media_d3)].copy()
print(f'Temperatura média: {temp_media_d3:.2f} | Registros: {len(alto_temp_d3)}'); alto_temp_d3.head()
```

### 7. Compare os conjuntos e explique o efeito da condição ambiental.

```python
reducao_d3=qtd_d3-len(alto_temp_d3); mantido_d3=len(alto_temp_d3)/qtd_d3*100 if qtd_d3 else 0
display(Markdown(f'''Havia **{qtd_d3}** registros acima do limiar; com temperatura acima da média, restaram **{len(alto_temp_d3)}**, redução de **{reducao_d3}**. Permaneceram **{mantido_d3:.2f}%** dos episódios. O segundo critério é mais restritivo e mostra coincidência entre alta demanda e temperatura acima da média, não causalidade.'''))
```

# Dataset 5 — Wind & Solar Energy Production (Kaggle)
## Situação
O operador deseja comparar períodos de alta produção solar e eólica, usando o máximo próprio de cada fonte.

### 1. Carregue a amostra e renomeie as variáveis para Geracao_Solar e Geracao_Eolica.

```python
ARQUIVO_D5 = '/content/drive/My Drive/SERS/EX_05_AMOSTRA.csv'

df5 = pd.read_csv(ARQUIVO_D5)
df5.columns = df5.columns.str.strip()

# Padroniza valores como "wind", "Wind" e possíveis espaços
df5['Source'] = (
    df5['Source']
    .astype(str)
    .str.strip()
    .str.title()
)

df5['Production'] = pd.to_numeric(
    df5['Production'],
    errors='coerce'
)

print('Colunas disponíveis:', df5.columns.tolist())
print('\nFontes encontradas:')
print(df5['Source'].value_counts())

# Separa as duas fontes
df5_solar = (
    df5[df5['Source'] == 'Solar']
    .copy()
    .rename(columns={'Production': 'Geracao_Solar'})
)

df5_eolica = (
    df5[df5['Source'] == 'Wind']
    .copy()
    .rename(columns={'Production': 'Geracao_Eolica'})
)

display(df5_solar.head())
display(df5_eolica.head())
```

### 2. Determine o máximo de cada fonte.

```python
max_solar_d5 = df5_solar['Geracao_Solar'].max()
max_eolica_d5 = df5_eolica['Geracao_Eolica'].max()

print(f'Máximo solar: {max_solar_d5:.2f}')
print(f'Máximo eólico: {max_eolica_d5:.2f}')
```

### 3. Calcule separadamente 70% de cada máximo.

```python
limiar_solar_d5 = 0.70 * max_solar_d5
limiar_eolica_d5 = 0.70 * max_eolica_d5

print(f'Limiar solar: {limiar_solar_d5:.2f}')
print(f'Limiar eólico: {limiar_eolica_d5:.2f}')
```

### 4. Crie DataFrames de alta geração solar e alta geração eólica.

```python
alto_solar_d5 = df5_solar[
    df5_solar['Geracao_Solar'] > limiar_solar_d5
].copy()

alto_eolica_d5 = df5_eolica[
    df5_eolica['Geracao_Eolica'] > limiar_eolica_d5
].copy()

display(alto_solar_d5.head())
display(alto_eolica_d5.head())
```

### 5. Conte os registros e calcule os percentuais.

```python
qtd_solar_d5 = len(alto_solar_d5)
qtd_eolica_d5 = len(alto_eolica_d5)

pct_solar_d5 = qtd_solar_d5 / len(df5_solar) * 100
pct_eolica_d5 = qtd_eolica_d5 / len(df5_eolica) * 100

print(
    f'Solar: {qtd_solar_d5} registros '
    f'({pct_solar_d5:.2f}%)'
)

print(
    f'Eólica: {qtd_eolica_d5} registros '
    f'({pct_eolica_d5:.2f}%)'
)
```

### 6. Compare qual fonte aparece mais vezes acima de 70% do próprio máximo.

```python
if qtd_solar_d5>qtd_eolica_d5: conclusao_d5='A fonte solar aparece com maior frequência.'
elif qtd_eolica_d5>qtd_solar_d5: conclusao_d5='A fonte eólica aparece com maior frequência.'
else: conclusao_d5='As duas fontes aparecem com a mesma frequência.'
display(Markdown(f'{conclusao_d5} Solar: **{qtd_solar_d5} ({pct_solar_d5:.2f}%)**; eólica: **{qtd_eolica_d5} ({pct_eolica_d5:.2f}%)**.'))
```

### 7. Explique por que o mesmo limite numérico não deve ser aplicado diretamente às duas fontes.

solar e eólica podem ter escalas, unidades e capacidades instaladas diferentes. Um limite absoluto poderia ser muito exigente para uma fonte e pouco exigente para outra. Usar 70% do máximo de cada variável ajusta o critério à escala observada. Como máximos podem ser sensíveis a valores extremos, uma análise posterior também poderia usar percentis.

# Dataset 6 — Individual Household Electric Power Consumption (UCI)

A residência deseja identificar demanda elevada que também apresente corrente acima da média.

### 1. Carregue a amostra e simplifique os nomes dos atributos.

### 2. Determine o máximo de potência ativa.

```python
max_d6=df6['Potencia_Ativa'].max(); print(f'Potência ativa máxima: {max_d6:.3f}')
```

### 3. Calcule 75% do máximo e filtre acima desse limite.

```python
limiar_d6=.75*max_d6; alto_d6=df6[df6['Potencia_Ativa']>limiar_d6].copy()
print(f'Limiar: {limiar_d6:.3f}'); alto_d6.head()
```

### 4. Calcule quantidade e percentual dos registros.

```python
qtd_d6=len(alto_d6); pct_d6=qtd_d6/len(df6)*100
print(f'Quantidade: {qtd_d6} | Percentual: {pct_d6:.2f}%')
```

### 5. Calcule a corrente média.

```python
media_corrente_d6=df6['Corrente'].mean(); print(f'Corrente média: {media_corrente_d6:.3f}')
```

### 6. Filtre simultaneamente potência ativa elevada e corrente acima da média.

```python
alto_corrente_d6=df6[(df6['Potencia_Ativa']>limiar_d6)&(df6['Corrente']>media_corrente_d6)].copy()
print(f'Registros no segundo recorte: {len(alto_corrente_d6)}'); alto_corrente_d6.head()
```

### 7. Compare os conjuntos e explique o efeito da corrente como segunda condição.

```python
reducao_d6=qtd_d6-len(alto_corrente_d6); mantido_d6=len(alto_corrente_d6)/qtd_d6*100 if qtd_d6 else 0
display(Markdown(f'''O primeiro filtro selecionou **{qtd_d6}** registros. Com corrente acima da média de **{media_corrente_d6:.3f}**, restaram **{len(alto_corrente_d6)}**, redução de **{reducao_d6}**. Assim, **{mantido_d6:.2f}%** dos episódios de potência elevada também tiveram corrente acima da média. A segunda condição torna o recorte mais específico.'''))
```
